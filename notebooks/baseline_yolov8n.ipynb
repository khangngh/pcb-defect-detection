# Baseline YOLOv8n trên Kaggle

Khung nhiệm vụ, chưa có code huấn luyện hoặc kết quả.


## Đầu vào và quy tắc

- Add Input đúng Kaggle Dataset Version của `PCB_Baseline_v1`.
- Giữ nguyên split và ID lớp; xem `data/README.md`.
- Tạo YAML theo đường dẫn input trong `/kaggle/working`.
- Cấu hình môi trường, GPU và huấn luyện trong notebook này khi bắt đầu nhiệm vụ.
- Chọn checkpoint bằng validation, chưa dùng test để tinh chỉnh.


## Nội dung cần bổ sung

1. Thư viện và phiên bản môi trường.
2. Đường dẫn dataset và kiểm tra input.
3. Seed và cấu hình huấn luyện.
4. Huấn luyện trên Kaggle và lưu checkpoint.
5. Kết quả validation, AP từng lớp và biểu đồ.
6. Link Kaggle Version và đường dẫn output.

Thống nhất ngân sách huấn luyện theo `docs/experiment_protocol.md` trước khi chạy.
